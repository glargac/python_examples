```python
def calculate_diameter_circle(radius:float):

    """
    With this fuction, is to calculate the diameter of a circle.

    Arguements:
        radius (float): is the radius of the circle.

    Return:
        float: diameter.
    """
    if(radius<0):
        return -1
    else:
        return radius * 2
```

```python
total_diameter_circle = calculate_diameter_circle(-6)

# 3. Mostramos el resultado
print(total_diameter_circle)
```
