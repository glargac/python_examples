```python
def make_sandwich(bread_type: str, filling: str, cheese: str = "none", toasted: bool = False) -> str:
    # 1. Handle the "toasted" prefix
    prefix = "Making a toasted" if toasted else "Making a"
    
    # 2. Handle the cheese part
    if cheese.lower() == "none":
        cheese_part = ""
    else:
        cheese_part = f" and {cheese} cheese"
    
    # 3. Combine everything into the final sentence
    return f"{prefix} {bread_type} sandwich with {filling}{cheese_part}."

# --- Testing for your specific output ---

# Output: Making a toasted wheat sandwich with turkey and cheddar cheese.
print(make_sandwich("wheat", "turkey", "ceddar", True))

# Output: Making a rye sandwich with ham.
print(make_sandwich("rye", "ham"))
    
```
